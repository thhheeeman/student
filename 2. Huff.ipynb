```python
import heapq

class Node:

    def __init__(self, freq, symbol, left=None, right=None):
        self.freq = freq
        self.symbol = symbol
        self.left = left
        self.right = right

    def __lt__(self, other):
        # Compare nodes based on frequency
        return self.freq < other.freq


def print_codes(node, code=""):

    # Leaf node
    if node.left is None and node.right is None:
        print(node.symbol, "->", code if code else "0")
        return

    if node.left:
        print_codes(node.left, code + "0")

    if node.right:
        print_codes(node.right, code + "1")


# Main Program
chars = ['a', 'b', 'c', 'd', 'e', 'f']
freq = [5, 9, 12, 13, 16, 45]

# Create leaf nodes
heap = [Node(f, c) for c, f in zip(chars, freq)]

# Build Min Heap
heapq.heapify(heap)

# Build Huffman Tree using Greedy Approach
while len(heap) > 1:

    left = heapq.heappop(heap)
    right = heapq.heappop(heap)

    # Create parent node
    parent = Node(
        left.freq + right.freq,
        left.symbol + right.symbol,
        left,
        right
    )

    heapq.heappush(heap, parent)


# Generate Huffman Codes
print("Huffman Codes:")
print_codes(heap[0])
```
