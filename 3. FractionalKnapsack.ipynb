```python
def fractional_knapsack(value, weight, capacity):

    n = len(value)
    ratio = []

    for i in range(n):
        ratio.append(value[i] / weight[i])

    for i in range(n):
        for j in range(i + 1, n):
            if ratio[i] < ratio[j]:
                ratio[i], ratio[j] = ratio[j], ratio[i]
                value[i], value[j] = value[j], value[i]
                weight[i], weight[j] = weight[j], weight[i]

    total = 0

    for i in range(n):
        if weight[i] <= capacity:
            total += value[i]
            capacity -= weight[i]
        else:
            total += value[i] * capacity / weight[i]
            break

    return total


n = int(input("Enter number of items: "))

value = list(map(int, input("Enter values: ").split()))
weight = list(map(int, input("Enter weights: ").split()))

capacity = int(input("Enter capacity: "))

print("Maximum value:", fractional_knapsack(value, weight, capacity))
```
