```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

# 1. Read dataset
df = pd.read_csv("stock_data.csv")

print("First five rows:")
print(df.head())

# 2. Preprocessing
df = df.dropna()

# 3. Convert date into date format
df["Date"] = pd.to_datetime(df["Date"])

# 4. Sort data by date
df = df.sort_values("Date")

# 5. Calculate next day stock return
df["Next_Return"] = (
    (df["Close"].shift(-1) - df["Close"])
    / df["Close"]
) * 100

df = df.dropna()

# 6. Select input and output
X = df[["Open", "High", "Low", "Close", "Volume"]]

y = df["Next_Return"]

# 7. Split dataset (80% training, 20% testing)
split = int(len(df) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]

# 8. Create Random Forest model
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

# 9. Train model
model.fit(X_train, y_train)

# 10. Predict stock returns
y_pred = model.predict(X_test)

# 11. Evaluate model
print("\nR2 Score:")
print(r2_score(y_test, y_pred))

print("\nRMSE:")
print(np.sqrt(mean_squared_error(y_test, y_pred)))

# 12. Display actual and predicted returns
print("\nActual Returns:")
print(y_test.head())

print("\nPredicted Returns:")
print(y_pred[:5])

# 13. Plot graph
plt.figure(figsize=(10, 5))

plt.plot(y_test.values, label="Actual Returns")
plt.plot(y_pred, label="Predicted Returns")

plt.xlabel("Days")
plt.ylabel("Stock Return (%)")

plt.title("Stock Return Prediction")

plt.legend()
plt.show()
```
